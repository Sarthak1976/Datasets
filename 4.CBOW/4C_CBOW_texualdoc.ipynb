```python
import gensim
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize, sent_tokenize

# 1. Data Preparation (Load text from a .txt file)
file_path = 'document.txt'  # Replace with your actual .txt filename

with open(file_path, 'r', encoding='utf-8') as f:
    text_data = f.read()

# Split text into sentences, then tokenize and lowercase each sentence into words
sentences = [word_tokenize(sentence.lower()) for sentence in sent_tokenize(text_data)]

# 2. Train Word2Vec CBOW Model (sg=0 specifies CBOW)
model = Word2Vec(sentences=sentences, vector_size=50, window=3, min_count=1, sg=0, epochs=10)

# 3. Evaluate Results (Check vocabulary and test word similarities)
print(f"Total vocabulary size: {len(model.wv.key_to_index)}")

test_word = 'data'  # Replace with a word actually present in your .txt file
if test_word in model.wv:
    similar_words = model.wv.most_similar(test_word, topn=3)
    print(f"Top words similar to '{test_word}':")
    for word, similarity in similar_words:
        print(f"  - {word}: {similarity:.4f}")
else:
    print(f"Word '{test_word}' not found in vocabulary.")
```
