```python
import pandas as pd
import gensim
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

# 1. Data Collection and Preprocessing
df = pd.read_csv('twitter_dataset.csv')
text_column = 'tweet' # Change this to your dataset's text column name

# Tokenize and lowercase text into a list of word lists (sentences)
sentences = [word_tokenize(str(text).lower()) for text in df[text_column].dropna()]

# 2. Train Word2Vec CBOW Model (sg=0 sets it to CBOW)
model = Word2Vec(sentences=sentences, vector_size=100, window=5, min_count=1, sg=0, epochs=10)

# 3. Evaluate Results (Check learned word embeddings / similarities)
test_word = 'happy' # Change to a word present in your tweets

if test_word in model.wv:
    similar_words = model.wv.most_similar(test_word, topn=5)
    print(f"Top words similar to '{test_word}':")
    for word, similarity in similar_words:
        print(f"  - {word}: {similarity:.4f}")
else:
    print(f"Word '{test_word}' not found in the vocabulary.")
```
